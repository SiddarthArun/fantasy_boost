```python
# IMPORTS
import pandas as pd
import nflreadpy as nfl
pd.set_option('display.max_rows', None)

df = nfl.load_player_stats(seasons=[2023,2024,2025], summary_level='week').to_pandas()
df[df['position'] == 'QB'].head(3)
```

```python
# FEATURE ENGINEERING

# Weighted Opportunities
df['wopr'] = (1.5*df['target_share']) + (0.7*df['air_yards_share'])

# RB Touches
df['touches'] = df['carries'] + df['targets']

# EWMA Scaled stats
ewma_metrics = ['fantasy_points_ppr', 'wopr', 'targets', 'receptions', 'receiving_yards', 'receiving_air_yards', 'receiving_yards_after_catch', 'receiving_first_downs', 'receiving_10', 'rushing_10','touches','receiving_20', 'target_share', 'air_yards_share', 'carries', 'rushing_yards', 'rushing_tds', 'rushing_first_downs', 'rushing_epa', 'completions', 'passing_yards', 'passing_tds', 'passing_interceptions', 'sacks_suffered', 'pacr', 'passing_epa', 'passing_cpoe', 'attempts']
for col in ewma_metrics:
    df[f'{col}_ewma3'] = df.groupby('player_id')[col].transform(lambda x: x.shift(1).ewm(span=4, min_periods=1).mean())

# Prior season performance
prior_season_avg = df.groupby(['player_id', 'season'])['fantasy_points_ppr'].mean().reset_index()
prior_season_avg['season'] += 1
prior_season_avg = prior_season_avg.rename(columns={'fantasy_points_ppr': 'prior_season_ppg'})
df = df.drop(columns=['prior_season_ppg'], errors='ignore')
df = df.merge(prior_season_avg, on=['player_id', 'season'], how='left')
df['prior_season_ppg'] = df['prior_season_ppg'].fillna(0)  # or fillna with league-average rookie WR output

# Boom/Bust likelihood
df['fantasy_points_ppr_std5'] = df.groupby('player_id')['fantasy_points_ppr'].transform(
    lambda x: x.shift(1).rolling(5, min_periods=2).std()
)

# Defense Stats
def_allowed = df.groupby(['opponent_team', 'season', 'week', 'position'])['fantasy_points_ppr'].sum().reset_index().sort_values(['opponent_team', 'position', 'season', 'week'])
def_allowed['def_allowed_l4w'] = def_allowed.groupby(['opponent_team', 'position'])['fantasy_points_ppr'].transform(lambda x: x.shift(1).rolling(4,1).mean())

df = df.drop(columns=['def_allowed_l4w'], errors='ignore')  # NEW — clears any stale version first
df = df.merge(def_allowed[['opponent_team', 'season', 'week', 'position', 'def_allowed_l4w']], on=['opponent_team', 'season', 'week', 'position'], how='left')

# Additions for RBs
```

```python
# MODEL SETUP
from sklearn.linear_model import LinearRegression
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

def rolling_eval(df, feature_cols, target_col, model_factory, test_season = 2025):
    results, preds_list = [], []
    weeks = sorted(df[df.season == test_season].week.unique())

    for week in weeks:
        train = df[(df.season < test_season) | ((df.season == test_season) & (df.week < week))]
        test = df[(df.season == test_season) & (df.week == week)]

        if len(test) < 5: continue

        model = model_factory()
        model.fit(train[feature_cols], train[target_col])
        preds = model.predict(test[feature_cols])

        results.append({'week': week, 'mae': mean_absolute_error(test[target_col], preds), 'n': len(test)})

        week_preds = test[['player_name', 'opponent_team', 'week']].copy()
        week_preds['pred'] = preds
        week_preds['actual'] = test[target_col].values
        preds_list.append(week_preds)
    
    return pd.DataFrame(results), pd.concat(preds_list, ignore_index=True)
```

## Wide Reciever

```python
feature_cols = ['wopr_ewma3', 'fantasy_points_ppr_ewma3', 'receiving_yards_after_catch_ewma3', 'def_allowed_l4w', 'fantasy_points_ppr_std5', 'prior_season_ppg']

wrs = df[df['position'] == 'WR']
wr_clean_df = wrs.dropna(subset=feature_cols + ['fantasy_points_ppr'])

wr_results, wr_preds = rolling_eval(wr_clean_df, feature_cols, 'fantasy_points_ppr', lambda: XGBRegressor(n_estimators=200, max_depth=4, learning_rate=0.05))
print(f"Overall MAE: {wr_results.mae.mean():.2f}")
```

## Takeaways
---
The best results thus far are with XGBoost with parameters: (n_estimators=200, max_depth=4, learning_rate=0.05). Optimal features thus far are: 
- ['touches_ewma3','target_share_ewma3','fantasy_points_ppr_ewma3','rushing_epa_ewma3','rushing_10_ewma3','def_allowed_l4w','fantasy_points_ppr_std5','prior_season_ppg'] for RBs
- ['wopr_ewma3', 'fantasy_points_ppr_ewma3', 'receiving_yards_after_catch_ewma3', 'def_allowed_l4w', 'fantasy_points_ppr_std5', 'prior_season_ppg'] for WRs.

The current workflow calculates predictions for players based on their stats over the past 3 games in such a way that performance 1 game prior carries slightly more worth than 2, 3, etc. It also factors in the strength of the defense they are going against and to some extent, how good the player has been historically. For example, if Ja'Marr Chase struggles the past 3 weeks, but sees an easy defense this current week, he will be projected to score a lot of points regardless given the strength of the defense and the fact that he is historically a very good player.
